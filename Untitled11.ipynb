```python
import cv2
import numpy as np
import imutils
from imutils import perspective
from scipy.spatial import distance as dist
from google.colab.patches import cv2_imshow
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
image_path = list(uploaded.keys())[0]
image = cv2.imread(image_path)

orig = image.copy()

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
gray = cv2.GaussianBlur(gray, (7, 7), 0)

edged = cv2.Canny(gray, 50, 100)
edged = cv2.dilate(edged, None, iterations=1)
edged = cv2.erode(edged, None, iterations=1)
```

```python
cnts = cv2.findContours(edged.copy(), cv2.RETR_EXTERNAL,
                        cv2.CHAIN_APPROX_SIMPLE)
cnts = imutils.grab_contours(cnts)

from imutils import contours
(cnts, _) = contours.sort_contours(cnts)
```

```python
def midpoint(ptA, ptB):
    return ((ptA[0] + ptB[0]) * 0.5, (ptA[1] + ptB[1]) * 0.5)
```

```python
pixelsPerMetric = None
object_count = 0
KNOWN_WIDTH = 2.0  # ₹1 coin diameter in cm

for c in cnts:
    if cv2.contourArea(c) < 500:
        continue

    object_count += 1

    box = cv2.minAreaRect(c)
    box = cv2.boxPoints(box)
    box = np.array(box, dtype="int")
    box = perspective.order_points(box)

    (tl, tr, br, bl) = box

    (tltrX, tltrY) = midpoint(tl, tr)
    (blbrX, blbrY) = midpoint(bl, br)

    (tlblX, tlblY) = midpoint(tl, bl)
    (trbrX, trbrY) = midpoint(tr, br)

    dA = dist.euclidean((tltrX, tltrY), (blbrX, blbrY))
    dB = dist.euclidean((tlblX, tlblY), (trbrX, trbrY))

    # Use first detected object (coin) as reference
    if pixelsPerMetric is None:
        pixelsPerMetric = dB / KNOWN_WIDTH

    dimA = dA / pixelsPerMetric
    dimB = dB / pixelsPerMetric

    cv2.drawContours(orig, [box.astype("int")], -1, (0, 255, 0), 2)

    cv2.putText(orig, "{:.2f}cm".format(dimA),
                (int(tltrX - 15), int(tltrY - 10)),
                cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 2)

    cv2.putText(orig, "{:.2f}cm".format(dimB),
                (int(trbrX + 10), int(trbrY)),
                cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 2)
```

```python
print("Total Objects Counted:", object_count)
cv2_imshow(orig)
```
